```python
import pandas as pd
import numpy as np
```

### UCSF_FSX7 Dataset

```python
ucsf = pd.read_csv('/Users/hieubui/Documents/Thesis 2027/data/ucsf_fsx7.csv')
ucsf.head()
```

```python
# Show information of all columns except 'ST*' columns
ucsf_info = ucsf.drop(columns=[col for col in ucsf.columns if col.startswith('ST')]).info()
print(ucsf_info)
print(ucsf.shape)
```

```python
# Drop unnecessary columns
ucsf1 = ucsf.drop(columns=['RID', 'VISCODE', 'IMAGEUID', 'EXAMDATE', 'RUNDATE',
                           'FSVER', 'OVERALLQC', 'TEMPQC', 'FRONTQC', 'PARQC',
                           'INSULAQC', 'OCCQC', 'BGQC', 'CWMQC', 'VENTQC', 'HIPPOQC', 'update_stamp', 'STATUS'])
```

```python
# Select only 3T values in FIELD_STRENGTH column
ucsf1 = ucsf1[ucsf1['FIELD_STRENGTH'] == '3T']

# Drop FIELD_STRENGTH column
ucsf1 = ucsf1.drop(columns=['FIELD_STRENGTH'])
ucsf1.head()
```

```python
# Sum missing values in each column
missing_values = ucsf1.isnull().sum()
print(missing_values.sort_values(ascending=False))
```

```python
# Drop features with most  missing values
ucsf1.drop(columns=['ST8SV', 'ST68SV'])
```

### adni_all Dataset

```python
all = pd.read_csv("/Users/hieubui/Documents/Thesis 2027/data/adni_all.csv")
all.head()
```

```python
# Drop unnecessary columns
all1 = all.drop(columns=['PTRACCAT', 'PTETHCAT', 'PTEDUCAT', 'AVDEL30MIN', 'AVDELTOT', 'LIMMTOTAL',
                             'VSWEIGHT', 'VSHEIGHT','VSBPSYS', 'VSBPDIA','NPISCORE', 'GENOTYPE'])
all1.info()
```

### Join ucsf1 and all dataset

```python
# Join the ucsf1 and all1 dataframes with 
# Matching ucsf1['PTID'] and all1[subject ID] and
# Matching ucsf1['VISCODE2'] and all1['visit']

merged_df = pd.merge(all1, ucsf1, left_on=['subject_id', 'visit'], right_on=['PTID', 'VISCODE2'], how='left')
merged_df.head()
```

```python
merged_info = merged_df.drop(columns=[col for col in merged_df.columns if col.startswith('ST')]).info()
print(merged_info)
print(merged_df.shape)
```

```python
merged_df = merged_df.drop(columns=['PTID', 'VISCODE2'])
```

### All_fMRI

```python
fmri = pd.read_csv("/Users/hieubui/Documents/Thesis 2027/data/all_fmri.csv")
fmri.head()
```

```python
# Select 3.0 in the fmri_field_str column
fmri = fmri[fmri['fmri_field_str'] == 3.0]

# Drop unnecessary columns
fmri1 = fmri.drop(columns=['image_id', 'study_id', 'fmri_date', 'fmri_mfr', 'fmri_mfr_model', 'fmri_field_str', 'fmri_sequence', 'fmri_flip_angle', 
                           'fmri_width', 'fmri_height', 'fmri_n_images', 'fmri_pixel_x', 'fmri_pixel_y', 'fmri_thickness', 'fmri_te'])
```

```python
print(fmri1.info())
print(fmri1.shape)
```

```python
# Count values in fmri_visit column
fmri1['fmri_visit'].value_counts()
```

```python
# Only keep rows having 'rsfmri' or 'Resting State fMRI' in fmri_description column
fmri1 = fmri1[fmri1['fmri_description'].isin(['rsfmri', 'Resting State fMRI'])]
```

```python
fmri1.head()
```

### Join fmri1 + merged_df

```python
# Join fmri1 with merged_df
# Matching both subject_id
# Matching fmri1['fmri_visit'] and merged_df['visit']
# Insert the fmri1 columns before merged_df['ST*'] columns
merged_fmri = pd.merge(merged_df, fmri1, left_on=['subject_id', 'visit'], right_on=['subject_id', 'fmri_visit'], how='left')
```

```python
fmri_cols = ['fmri_visit', 'fmri_description', 'fmri_tr']

# Find first ST column
st_pos = next(i for i, c in enumerate(merged_fmri.columns) if c.startswith('ST'))

# Move the 3 fMRI columns right before ST*
for col in reversed(fmri_cols):
    merged_fmri.insert(st_pos, col, merged_fmri.pop(col))
```
